Q2. What is the difference between the functions loc and iloc in pandas.DataFrame?

Ans:- loc[] and iloc[] are used for selecting data from a Pandas DataFrame, The loc[] function selects rows and columns using their labels or names, whereas iloc[] selects rows and columns using integer index positions, For example, df.loc[2] accesses the row whose label is 2, while df.iloc[2] accesses the third row of the DataFrame regardless of its label, Therefore, loc[] is label-based indexing and iloc[] is position-based indexing.

Q4. Write a code to find the following statistical measurements for the above dataframe df1:
    
    (i) mean of each and every column present in the dataframe.

    (ii) standard deviation of column, ‘column_2’

```python
import pandas as pd
import numpy as np

columns = ['column_1','column_2','column_3',
           'column_4','column_5','column_6']

indices = [1,2,3,4,5,6]

df1 = pd.DataFrame(
    np.random.rand(6,6),
    columns=columns,
    index=indices
)

# Mean of all columns
print(df1.mean())

# Standard deviation of column_2
print(df1['column_2'].std())
```

Q6. What do you understand about the windows function in pandas and list the types of windows
functions?

Ans:- Window functions in Pandas are used to perform calculations over a set of rows related to the current row, These functions are commonly used in time-series analysis, moving averages, cumulative statistics, and data smoothing, Instead of calculating values on the entire dataset, window functions operate on a specific window of observations around each row.

The main types of window functions in Pandas are:

1. Rolling Window Functions

2. Weighted Window Functions

3. Expanding Window Functions

4. Exponentially Weighted Moving Window Functions (EWM)

Example:-

```python
df.rolling()
df.expanding()
df.ewm()
```

Q8. Write a Python program that takes in two dates as input (in the format YYYY-MM-DD) and
calculates the difference between them in days, hours, and minutes using Pandas time delta. The
program should prompt the user to enter the dates and display the result.

```python
import pandas as pd

date1 = input("Enter first date (YYYY-MM-DD): ")
date2 = input("Enter second date (YYYY-MM-DD): ")

date1 = pd.to_datetime(date1)
date2 = pd.to_datetime(date2)

difference = date2 - date1

print("Days:", difference.days)
print("Hours:", difference.total_seconds()/3600)
print("Minutes:", difference.total_seconds()/60)
```

Q9. Write a Python program that reads a CSV file containing categorical data and converts a specified
column to a categorical data type. The program should prompt the user to enter the file path, column
name, and category order, and then display the sorted data.

```python
import pandas as pd

file_path = input("Enter CSV file path: ")
column_name = input("Enter column name: ")

order = input(
    "Enter category order separated by commas: "
).split(',')

df = pd.read_csv(file_path)

df[column_name] = pd.Categorical(
    df[column_name],
    categories=order,
    ordered=True
)

df_sorted = df.sort_values(column_name)

print(df_sorted)
```

Q10. Write a Python program that reads a CSV file containing sales data for different products and
visualizes the data using a stacked bar chart to show the sales of each product category over time. The
program should prompt the user to enter the file path and display the chart.

```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = input("Enter CSV file path: ")

df = pd.read_csv(file_path)

df.plot(
    x='Date',
    kind='bar',
    stacked=True,
    figsize=(10,6)
)

plt.title("Sales by Product Category")
plt.ylabel("Sales")
plt.show()
```
